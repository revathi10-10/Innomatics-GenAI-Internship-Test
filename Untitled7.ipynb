```python
import pandas as pd

# Load final dataset
df = pd.read_csv("/content/final_food_delivery_dataset.csv")

# Filter only GOLD members
gold_df = df[df['membership'] == 'Gold']

# Calculate total revenue city-wise
city_revenue = (
    gold_df
    .groupby('city')['total_amount']
    .sum()
    .sort_values(ascending=False)
)
print("total_amount by city(Gold members):")
print(city_revenue)
# Group by cuisine and calculate the average order value
avg_order_cuisine = final_df.groupby('cuisine')['total_amount'].mean().sort_values(ascending=False)

print("\nAverage Order Value by Cuisine:")
print(avg_order_cuisine)
```

```python
# Group by user_id and sum their total_amount
user_spending = final_df.groupby('user_id')['total_amount'].sum()

# Filter for users who spent more than 1000
high_spenders = user_spending[user_spending > 1000]

# Count how many distinct users meet the criteria
distinct_high_spenders_count = high_spenders.count()

print(f"Number of distinct users with total spend > 1000: {distinct_high_spenders_count}")
```

```python
import pandas as pd

# Define the rating bins and labels based on your test options
bins = [3.0, 3.5, 4.0, 4.5, 5.0]
labels = ['3.0-3.5', '3.6-4.0', '4.1-4.5', '4.6-5.0']

# Create the rating_range column
final_df['rating_range'] = pd.cut(final_df['rating'], bins=bins, labels=labels, include_lowest=True)

# Calculate total revenue for each range
rating_revenue = final_df.groupby('rating_range')['total_amount'].sum().sort_values(ascending=False)

print("Total Revenue by Rating Range:")
print(rating_revenue)
```

```python
# Calculate total number of orders
total_orders = len(final_df)

# Calculate number of orders from Gold members
gold_orders = len(final_df[final_df['membership'] == 'Gold'])

# Calculate percentage and round to nearest integer
gold_percentage = round((gold_orders / total_orders) * 100)

print(f"Percentage of orders by Gold members: {gold_percentage}%")
```

```python
# Group by cuisine to count unique restaurants and sum revenue
cuisine_stats = final_df.groupby('cuisine').agg(
    distinct_restaurants=('restaurant_id', 'nunique'),
    total_revenue=('total_amount', 'sum')
).sort_values(by='distinct_restaurants')

print("Cuisine Stats (Sorted by Number of Restaurants):")
print(cuisine_stats)
```

```python
# Group by restaurant name to get count and average price
rest_stats = final_df.groupby('restaurant_name').agg(
    total_orders=('order_id', 'count'),
    avg_order_value=('total_amount', 'mean')
)

# Filter for restaurants with less than 20 total orders
filtered_rest = rest_stats[rest_stats['total_orders'] < 20]

# Sort to find the highest average order value
highest_avg_small_rest = filtered_rest.sort_values(by='avg_order_value', ascending=False)

print("Highest Avg Order Value (with < 20 orders):")
print(highest_avg_small_rest.head())
```

```python
# Group by both membership level and cuisine
combo_revenue = final_df.groupby(['membership', 'cuisine'])['total_amount'].sum().sort_values(ascending=False)

print("\nRevenue by Membership + Cuisine Combination:")
print(combo_revenue.head())
```
